```
import urllib.request, traceback
from pathlib import Path
from datetime import datetime, timezone
code_url = "https://raw.githubusercontent.com/ankurchemidyes/ankur-website/2174ff02ab70422233c70fd6806277c560d1221f/btc-research/btc_15m_1h_match_2014_2026_v5_resume_status.py"
try:
    source = urllib.request.urlopen(code_url, timeout=90).read().decode('utf-8')
    exec(compile(source, 'btc_15m_1h_match_2014_2026_v5_resume_status.py', 'exec'))
except Exception:
    details = traceback.format_exc()
    print('\nERROR FOUND. Full details will be written into My Drive when mounted.\n')
    print(details[-4500:])
    output_root = Path("/content/drive/MyDrive/BTC_15MIN_1H_2014_2026_V3_RESEARCH")
    if output_root.is_dir():
        (output_root / 'LAST_ERROR.txt').write_text(details, encoding='utf-8')
        (output_root / 'RUN_STATUS.txt').write_text('ERROR\nCheck LAST_ERROR.txt\n' + str(datetime.now(timezone.utc)), encoding='utf-8')
    raise
```
