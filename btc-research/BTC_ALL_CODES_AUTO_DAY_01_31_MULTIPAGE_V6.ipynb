# BTC — सभी पैटर्न कोड की मल्टीपेज PDF में तारीख का दिन 01–31 क्रम से (V6)
**कोई Choose File, कोड इनपुट, या तारीख इनपुट नहीं।** यह अपने-आप पहले से बने V5 `PDFS_GROUPED/UTC_23_00` और `PDFS_GROUPED/UTC_11_12` में मौजूद **हर** कोड-PDF को पढ़कर उसके पेजों को केवल UTC तारीख के **दिन (01, 02, ..., 31)** के क्रम में लगाएगा। **साल और महीना नहीं पूछेगा।**
एक कोड की **एक ही मल्टीपेज PDF** बनी रहेगी; बाकी सभी पेज भी उसी फाइल में बने रहेंगे। उदाहरण: `23gggggg00.pdf` के सभी 23 तारीख वाले रिकॉर्ड एकसाथ रखे जाएँगे, बाकी दिन-नंबर वाले पेज क्रम में आएँगे। तारीखों के बीच साल-महीना अलग होने पर भी दिन-नंबर के आधार पर समूह बनेंगे; प्रत्येक पेज पर पूरी असली UTC तारीख बनी रहेगी।
**मूल सुरक्षित:** V5 की मूल PDF नहीं बदलेगी, नहीं हटेगी। नई फाइलें `MyDrive/BTC_MATCHED_CODES_MULTIPAGE_V5/ALL_CODES_DAY_01_TO_31_SORTED_V6/UTC_23_00/` और `UTC_11_12/` में बनेंगी। `REPORTS/FINAL_SUMMARY.txt` में कितनी PDF और कितने पेज सफलतापूर्वक बने; `ALL_CODES_PAGE_COUNTS.csv` में प्रति-कोड पेज की संख्या और `ISSUES_REQUIRING_ATTENTION.csv` में किसी त्रुटि की जानकारी।
**पहले V5 जरूरी है।** यदि `PDFS_GROUPED` में पुरानी मल्टीपेज PDF नहीं हैं तो इस स्क्रिप्ट से नई OHLC डेटा डाउनलोड या नए चार्ट नहीं बनेंगे; पहले V5 वाला Colab चलाना होगा। यहाँ केवल पेजों को नए क्रम में कॉपी किया जाता है।
