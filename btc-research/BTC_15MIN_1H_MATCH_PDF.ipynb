# Bitcoin 15M / 1H — 2014–2026 historical matched-signal PDF research
Run the **single Python cell** below and authorize Google Drive. It downloads market OHLC data and creates **one** 15-minute PDF group folder: `MyDrive/BTC_15MIN_1H_MATCH_PDF_V2_ORIGINAL_STYLE/ONE_15MIN_PDF_SET`.
PDF group name example: `15RRG084523.pdf`. One PDF contains all matching historical chart pages having that 15-minute code, UTC time and day, across months/years.
Every page displays a 15-minute chart above a one-hour chart, arrows at SuperTrend (1,1) flips, CRSI (3,2,100), both signal codes and full UTC dates. Only **same-direction 1H/15M flip matches** generate PDF pages. All 15M signals including nonmatches remain in CSV.
For a one-hour 08:00 UTC signal, the accepted 15-minute signal times are 07:00 through 08:45 UTC inclusive. **Historical matching is not proof of realtime signal availability.**
**Integrity limits:** Bitstamp BTC/USD is used before Binance BTC/USDT and the markets differ. This notebook has **not** been execution-tested in Colab or independently confirmed to match an exact TradingView ticker/indicator implementation.


```
import urllib.request
url = "https://raw.githubusercontent.com/ankurchemidyes/ankur-website/btc-15m-pdf-colab/btc-research/btc_15m_1h_match_pdf.py"
source = urllib.request.urlopen(url, timeout=60).read().decode('utf-8')
exec(compile(source, 'btc_15m_1h_match_pdf.py', 'exec'))
```
