# Bitcoin 2013 से आज तक — साफ OHLCV डेटा (बिना इंडिकेटर)
**यह शुद्ध डेटा-एक्सपोर्ट है**। इसमें RSI, SuperTrend, EMA, MACD, Entry/Exit, ट्रेडिंग सिग्नल या कोई ऑर्डर नहीं है।
**स्रोत:** एक ही एक्सचेंज Bitstamp, मार्केट **BTC/USD**। Bitstamp के सार्वजनिक OHLC API से असली **15-मिनट OHLCV** डेटा लिया जाता है; 1 घंटे और 1 दिन की कैंडलें तभी बनती हैं जब उनके क्रमशः सभी 4 या 96 मूल 15-मिनट कैंडल उपलब्ध हों।
**अवधि:** 1 जनवरी 2013 से इस Run के समय तक की आखिरी *पूरी हो चुकी* 15-मिनट कैंडल (UTC)। जहाँ मूल डेटा उपलब्ध नहीं होता या असंगत है, उसे **छिपाया/बनाया नहीं जाता**; `AUDIT_REPORTS` फ़ोल्डर में कमी व गलत रिकॉर्ड रिपोर्ट होते हैं। पूरा डेटा उपलब्ध होने की गारंटी नहीं है।
**Google Drive:** `MyDrive/BTC_ORIGINAL_OHLCV_2013_NOW_V1/`। `CLEAN_CSV` में पूरे 15-मिनट डेटा की CSV.gz, हर वर्ष की 15M CSV.gz, पूर्ण घंटे/दिन की CSV.gz। `AUDIT_REPORTS` में गुणवत्ता की रिपोर्ट और गायब कैंडलें।
**मोबाइल:** नीचे **एक ही ▶ Run** दबाएँ और Drive की अनुमति दें। लंबे डाउनलोड के दौरान `RUN_STATUS.txt` तथा `DOWNLOAD_PARTS` में प्रगति सुरक्षित होती है; वही Run दोबारा करने पर डाउनलोड पूरा होने की जगह से आगे बढ़ेगा।
**नोट:** Bitstamp BTC/USD और Binance BTC/USDT अलग बाजार हैं। यह मूल सार्वजनिक कैंडल डेटा है, मूल ट्रेड-टिक रिकॉर्ड नहीं; न ही किसी TradingView चार्ट से स्वतः प्रमाणित समानता।
